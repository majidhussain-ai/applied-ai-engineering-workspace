```python
# Task Management: There are two methods to implement task manager 
# 1. asyncio.create_task()
# 2. async with asyncio.TaskGroup() as allies or tg: (this method is industry standard)

async def getting_db_query(user_id):
    print(f"Connecting to database with ID {user_id} to reasoning...")
    await asyncio.sleep(2)
    return f"Response is ready for user_id: {user_id} "

async def downloading_data_website(data):
    print(f"Opening website for downloading: {data} ")
    await asyncio.sleep(3)
    return f"Data is ready {data} for downloadable."

async def generating_embedding(chunks):   #This function is just for create_task
    print(f"Generator model is ready to generate embeding for '{chunks}'")
    await asyncio.sleep(4)
    raise ConnectionError('Server is slow down to converting in vectors.')


async def generating_embeddings(chunks):   #This function is just for TaskGroup
    try:
        print(f"Generator model is ready to generate embeding for '{chunks}'")
        await asyncio.sleep(5)
        print("For chunks Vectors (numeric) is ready for database")
    except asyncio.CancelledError:
        print('Remaining all working functions is hide for just TaskGroup')


```

```python
"""
Industry grade Task Manager:
If one task is crash other tasks will automatically skip their work to 
prevent to leak source or waste of memory
"""

async def executing_pipeline():
    embeding_chunks = [
        'Hy, Majid Congratulations to completing python',
        'Welcome in Advance Production grade python',
        'Youre curious to learn new concepts to becoming an Applied Orchestrator AI Engineer.'
    ]

    async with asyncio.TaskGroup() as tg:
        try:
            db_query = tg.create_task(getting_db_query(102))
            download_web_data = tg.create_task(downloading_data_website('majid_1135.png'))
            generate_embed = tg.create_task(generating_embedding(embeding_chunks))
            generate_embeds = tg.create_task(generating_embeddings(embeding_chunks))
        
        except ExceptionGroup as eg:
            print(f"Task is successfully stop working for error of: {eg}")

await executing_pipeline()
```

```python
# Taskgroup with multiple coroutine functions


async def getting_db_query(user_id):
    print(f"Connecting to database with ID {user_id} to reasoning...")
    await asyncio.sleep(2)
    return f"Response is ready for user_id: {user_id} "

async def downloading_data_website(data):
    print(f"Opening website for downloading: {data} ")
    await asyncio.sleep(3)
    return f"Data is downloaded in form of {data} ."

async def generating_embedding(chunks):   #This function is just for create_task
    print(f"Generator model is ready to generate embeding for '{chunks}'")
    await asyncio.sleep(4)
    return f"Generator model is ready to convert text into numeric vectors"


async def generating_embeddings(chunks):   #This function is just for TaskGroup

    print(f"Generator model is ready to generate embeding for '{chunks}'")
    await asyncio.sleep(5)
    return "For chunks Vectors (numeric) is ready for database"

async def executing_pipeline():
    embeding_chunks = [
        'Hy, Majid Congratulations to completing python',
        'Welcome in Advance Production grade python',
        'Youre curious to learn new concepts to becoming an Applied Orchestrator AI Engineer.'
    ]

    async with asyncio.TaskGroup() as tg:
        pipeline = {
            "db_query" : tg.create_task(getting_db_query(102)),
            "download_web_data" : tg.create_task(downloading_data_website('majid_1135.png')),
            "generate_embed" : tg.create_task(generating_embedding(embeding_chunks)),
            "generate_embeds" : tg.create_task(generating_embeddings(embeding_chunks))
                    
        }

    print(f"DataBase Fetched Result: {pipeline['db_query'].result()}")
    print(f"Data is ready for downloadable: {pipeline['download_web_data'].result()}")

await executing_pipeline()
```

```python
# usig httpx to getting data through url from online any platform or websites

async def perform_get_request():
    url = "https://httpbin.org/get"
    # url = "https://httpbin.org"    #This will give jsondecode error

    async with httpx.AsyncClient() as client:
        response = await client.get(url, params={'query': 'Applied AI', 'limit':1})
        print(f"Status code: {response.status_code}")

        return response.json()

get_data = await perform_get_request()
print(f"Fetched url: {get_data['url']}")
```

```python
# main executable pipeline 

async def get_llm_response(client, url):
    response = await client.get(url)
    return (url ,response.status_code)

async def main():
    urls = ['https://github.com/majidhussain-ai', 'https://httpbin.org/get']
    async with httpx.AsyncClient() as client:
        tasks = [get_llm_response(client,u) for u in urls]
        results = await asyncio.gather(*tasks, return_exceptions=True)
        for result in results:
            if isinstance(result, BaseException):
                print(f"Request failed: {result}")
                continue
            url, status = result
            print(f"Get URL: {url}\nGet status: {status} ")

await main()
```

```python

# posting data to a web server

async def perform_post_request():
    url = "https://httpbin.org/post"
    ai_payload = {
        "model": "GPT-4o",
        "message": [{"role": "user", "content": "Explain AsyncioAI"}],
    }

    async with httpx.AsyncClient() as client:
        response = await client.post(url, json=ai_payload)
        response.raise_for_status()
        print(f"Status Code: {response.status_code}")
        return response.json()

post_data = await perform_post_request()
print(f"Sent payload received back: {post_data['json']}")
```
