```python
# Working with json and csv data to store config , parameters of RAG agent, fine tuning in disk file with serialize
import logging
logger = logging.getLogger('EnsureFileSaveDisk')
logging.basicConfig(level=logging.INFO)

def save_agent_config(experiment_payload, output_dir, file_name):
    output_dir.mkdir(parents = True, exist_ok = True)
    target_file = output_dir/f"{file_name}.json"
    with target_file.open(mode = 'w', encoding = 'utf-8') as stream_file:
        json.dump(experiment_payload, stream_file, indent=2 , ensure_ascii=False)
    logger.info(f"[Saved Config] {target_file.name} (seized {(target_file.stat().st_size)/1024 : .3f} MBs)")

    return target_file


def load_agent_data(file_path):
    with file_path.open(mode = 'r', encoding = 'utf-8') as stream_file:
        return json.load(stream_file)

experiment_directory = Path('./ai_experiment').resolve()
agent_metadata = {
    "model_name": "gemini-2.5-pro",
    "temperature": 0.2,
    "max_output_tokens": 2048,
    "system_instruction": "Aap ek professional AI assistant hain. Urdu aur English dono support karein.",
    "tools_enabled": ["rag_retriever", "code_interpreter"]
}

saved_path = save_agent_config(agent_metadata, experiment_directory, 'agent_config_v1')
loaded_config = load_agent_data(saved_path)
logger.info(f'[Loaded Instruction] {loaded_config['system_instruction']}')
```

```python
import json
import logging
from pathlib import Path
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger('FineTuneRecoredStream')

def appends_fine_tune_records(dataset_path, records):
    dataset_path.mkdir(parents = True, exit_ok = True)
    with dataset_path.open(mode = 'a', encoding = 'utf-8') as stream_line:
        for record in records:
            line = json.dumps(record, ensure_ascii=False)
            stream_line.write(line + "\n")
    logging.info(f"[Appended Records] {len(records)} records in {dataset_path.name}")

def stream_record(dataset_path):
    with dataset_path.open(mode='w', encoding = 'utf-8') as stream_line:
        for line_num, line in enumerate(stream_line):
            cleaned = line.strip()
            if cleaned:
                yield json.loads(cleaned)

dataset_file = Path("./fine_tune_data/instrant_chat.jsonl").resolve()
new_training_samples = [
    {
        "messages": [
            {"role": "user", "content": "What is Quantization in deep learning?"},
            {"role": "assistant", "content": "Quantization reduces weight precision from FP32 to INT8/INT4 to reduce VRAM."}
        ]
    },
    {
        "messages": [
            {"role": "user", "content": "Explain Vector Embedding."},
            {"role": "assistant", "content": "A high-dimensional numerical representation of textual semantic meaning."}
        ]
    }
]

# write appends \
appends_fine_tune_records(dataset_file, new_training_samples)

# reading line by line 
for sample in stream_record(dataset_file):
    user_query = sample['content'][0]['message']
```
