```python
def run_pipeline(*steps):
    for step in steps:
        step()
def load_data():
        print('Data is loading....')
def create_embedings():
        print('Embeding is creating....') 
def retrieve_documents():
        print('Documents is retrieving....') 
execute = run_pipeline(
    load_data,
    create_embedings,
    retrieve_documents
)
print(execute)
```

```python
from typing import Callable, Any


def run_pipeline(*steps: Callable[[Any], Any]) -> Any:
    """
    Execute pipeline steps sequentially.

    Each step receives the output of the previous step.
    """

    data = None

    for step in steps:
        data = step(data)

    return data


def load_data(_: Any) -> list[str]:
    """Load documents from the data source."""
    
    documents = [
        "Document 1",
        "Document 2",
        "Document 3",
    ]

    print("Data loaded.")
    return documents


def create_embeddings(documents: list[str]) -> list[str]:
    """Create embeddings for the documents."""

    embeddings = [
        f"Embedding({document})"
        for document in documents
    ]

    print("Embeddings created.")
    return embeddings


def retrieve_documents(embeddings: list[str]) -> list[str]:
    """Retrieve relevant documents using embeddings."""

    retrieved_documents = embeddings[:2]

    print("Documents retrieved.")
    return retrieved_documents


result = run_pipeline(
    load_data,
    create_embeddings,
    retrieve_documents,
)

print("\nFinal result:")
print(result)
```

```python
# Hissa 1: Target Function (Asli kaam karne wala)
def target_function(name, age, model="default", temperature=0.7):
    """
    Yeh function inputs leta hai aur unhein pipe (|) 
    ke sath jod kar aik single string return karta hai.
    """
    return f"{name} | {age} | {model} | {temperature}"


# Hissa 2: Wrapper Function (Darmiyan wala waiter)
def wrapper(*args, **kwargs):
    """
    Yeh function har qism ka data accept karta hai (*args aur **kwargs ke zariye)
    aur usay bina tabdeel kiye target_function ko forward kar deta hai.
    """
    return target_function(*args, **kwargs)


# Hissa 3: Function Call (Execution)
# Hum ne wrapper ko call kiya aur data pass kiya
result = wrapper(
    "Majid",
    22,
    model="LLM",
    temperature=0.2
)

# Output ko screen par print karna
print(result)
```

```python
# Mini exercise about more advance functions 
def process_text(*text, lowercase = True, prefix = "[Processed ]"):
    docs = []
    for txt in text:
        data = txt.strip()
        if lowercase:
            data = data.lower()
        docs.append(data)
    return f'{prefix}{docs}'


result = process_text(
    "   HELLO WORLD   ",
    ' Hy amjad what are you doing ',
    "I'm here in parks waiting for you since 2 hours ",
    lowercase=True,
    prefix="[CLEANED] "
)

print(result)
    
```

## Higher order functions:
- Take functions as an argument(input)
- Return functions as a new function

```python
# Ever simple example
def play_punjabi_music():
    return ('Gabru nami song started....')

def play_urdu_music():
    return ('Kaise ye howa khasara nami song started....')

def music_play(songs_function):
    print('Song is being started...')
    print(songs_function())
    print('Song is being end now closed.\n')

music_play(play_punjabi_music)
music_play(play_urdu_music)
```

```python
# simple example
def show_info(*args, **kwargs):
    for name in args:
        print(name)
    for key , value in kwargs.items():
        print(f'{key} -> {value}')
my_func = show_info
my_func(
    'majid hussain',
    22,
    'single',
    sems = 'graducation',
    uni_name = 'IUB',
    grad = 'B'
)
```
