## Logging:
- logging is production standard to control the flow of application and easy to get error from any application by using logging file.

`Logging have 5 levels:`
* logging.degub()
* logging.info()
* logging.warning()
* logging.error()
* logging.critical()

```python
# simple examples with configuration
import logging

logging.basicConfig(
    level=logging.DEBUG     #very low level
)
logger = logging.getLogger('AIPipeline')
logger.debug("This is just debuging message")
logger.info("This is just info message")
logger.warning("This is just warning message")
logger.error("This is just error message")
logger.critical("This is critical error to handle")
```

```python
# creating log file to save error there as like production standard
from datetime import datetime
import logging
logging.basicConfig(
    filename='App.log',
    level= logging.DEBUG,
    format='%(asctime)s | %(levelname)s | %(name)s | %(message)s',
    datefmt="%Y-%m-%d %H:%M:%S"
)

logger = logging.getLogger(__name__)

logger.info("This is code info message to successfully done this or that work")
logger.warning("This is code for warning message to avaid from memory full space")
logger.error("This is code error message to shut down server or db connection erro")
logger.critical("This is code for critical message to Internet cut off on whole world")
```

```python
# logging with function handling
import logging
logging.basicConfig(
    level=logging.INFO,
    format="[%(asctime)s] | %(levelname)s | %(name)s | %(message)s",
    datefmt="%Y-%m-%d %H:%M:%S"
)

logger = logging.getLogger(__name__)

def process_document(data_list):
    logger.info("Started Processing Documents...")
    if not data_list:
        logger.warning("Documents is empty not processed.")
        return None

    
    logger.info(f"Total {len(data_list)} items are started to process.")
    return len(data_list)

process_document([12, 23, 23,2, 12, 2,4])
process_document([])
```

```python
# logger.exception: this is only useful inside except block to handle exact function name error

import logging
logging.basicConfig(
    level=logging.INFO,
    format="[%(asctime)s] | %(levelname)s | %(name)s | %(message)s ",
    datefmt= "%Y-%m-%d %H:%M:%S"
)

logger = logging.getLogger(__name__)

def average_calculate(total_numbers, students):
    try:
        logger.info(f"Calculating average marks total marks: {total_numbers} and students {students}")
        result = total_numbers/students
        return result
    except ZeroDivisionError as e:
        logger.exception(f"Failed to calculate average due to division by zero")
        return None

average_calculate(450,30)
average_calculate(500, 0)
```
