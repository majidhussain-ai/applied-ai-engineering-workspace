## Encapsulation:
- Encapsulation is the bundling of data and methods into a single class while restricting direct external access to internal states.
- Bundling: Data + Methods in One Class.
- Data Hiding: Protecting variables using private/protected prefixes (_ or __)

- In commands like (update, save, append) we don't need to write return keyword
- In Qurries state like (.get, .search, .calculate anything) we need to return performed actions.

```python
class Demonostration: 
    def __init__(self, name, age):
        self._name = name
        self.__age = age

    @property
    def access(self):
        return f"My name is {self._name} and my age is {self.__age}"

    @access.setter 
    def access(self, new_age):
        if new_age>0:
            self.__age = new_age
            return f"My name is {self._name} with modified age is {self.__age}"
        else:
            print("Age can't be negative or zero")

enc = Demonostration('majid', 22)
print(enc.access)

enc.access = 23 

print(enc.access)
```

## Aggregation in OOPs:
- In Aggregation 'Has-A' relationship between two classes, one class object pass through another class.
- Class where object passed of anyother class has access or modify their attributes and methods but these two are completely independent.
- Deletion of one class object doesn't effect of another class object or methods and attributes.
Example: A university has a teacher , but in some cases university is closed teacher will teacher in other academic but not completely destroy their knowledge.

```python
# Single inheritance code example
class BaseChatHistory:
    def __init__(self, session_id):
        self.session_id = session_id
        self.messages = []

    def add_message(self, role, content):
        self.messages.append({'role' : role , 'content' : content})

class ConversationSummeryMemory(BaseChatHistory):
    def get_formated_history(self):
        formated = [f" {msg['role'].upper()} : {msg['content']} " for msg in self.messages]
        return '\n'.join(formated)

memory = ConversationSummeryMemory('user_101')
memory.add_message('user' , 'What is Vector DataBase')
memory.add_message('assistant', 'Vector Data store high dimensional of semantic meaning to retrive fast from similar meaning')
print(memory.get_formated_history())
```

```python
# Single Inheritance
class BaseInputValidator:
    def __init__(self, max_tokens = 500):
        self.max_tokens = max_tokens

    def is_valid_length(self, text):
        return len(text) <= self.max_tokens

class SafetyGuardrails(BaseInputValidator):
    def __init__(self, max_tokens = 500):
        super().__init__(max_tokens)
        self.banned_words = ['exploid', 'bypass', 'hack']

    def sanitize(self, prompt):
        if not self.is_valid_length(prompt):
            raise ValueError('Prompt exceeds allowed tokens/characters length')

        cleaned_prompt = prompt
        for text in self.banned_words:
            cleaned_prompt = cleaned_prompt.replace(text, 'REDACTED')
        return cleaned_prompt

guard = SafetyGuardrails(200)
safe_text = guard.sanitize('How to hack system security')
print(safe_text)
```

```python
# Chunks splitter
my_str = 'ABCDEFGHIJKLMNO'
chunks = 5
over_lap = 2
step = chunks - over_lap
def character_spliter(document):
    return [document[char : char + chunks] for char in range(0, len(document) , step)]

char_split = character_spliter(my_str)
print(char_split)
```

```python
# Hierarchi inheritance the most useful concept and example
# Chunking and spliting Document before give to llms or vector databases

class BaseTextSplitter:
    def __init__(self, chunk_size, chunk_overlaps):
        self.chunk_size = chunk_size
        self.chunk_overlaps = chunk_overlaps

    def validate_chunks(self, chunks):
        return f"Successfully created {len(chunks)} chunks (Overlap {self.chunk_overlaps})"

class CharacterSplitter(BaseTextSplitter):
    def split(self, documents):
        steps = self.chunk_size - self.chunk_overlaps
        return [documents[char : char + self.chunk_size] for char in range(0, len(documents), steps)]

class WordSplitter(BaseTextSplitter):
    def split(self, documents):
        words = documents.split()
        return [" ".join(words[i : i + self.chunk_size]) for i in range(0, len(words), self.chunk_size)]

raw_doc = "Artificial Intelligence is transforming how software systems are built and deployed in 2026."

char_worker = CharacterSplitter(chunk_size=27, chunk_overlaps=4)
word_worker = WordSplitter(chunk_size=5, chunk_overlaps=0)

char_split = char_worker.split(raw_doc)
word_split = word_worker.split(raw_doc)

print(char_split)
print(word_split)

print(char_worker.validate_chunks(char_split))
print(word_worker.validate_chunks(word_split))
```

```python
# Ex3: Method overriding: Cache in-memory

class DirectEmbeddingService:
    def get_embedding(self, text):
        print(f'[Network API Calling] Generating Embedding for: {text}')

        return [0.2, 0.4 , 0.5, 0.5]

class CachedEmbeddingService(DirectEmbeddingService):
    def __init__(self):
        super().__init__()
        self._chache = {}

    def get_embedding(self, text):
        clean_key = text.strip().lower()
        if clean_key in self._chache:
            print(f"[Hit Cache] Returning embedding from memory for: {text}")
            return self._chache[clean_key]

        # if prompt embedding not found in chache memory then call parent class api 
        vector = super().get_embedding(text)       #method override

        # store parent class embedding in chache for future response
        self._chache[clean_key] = vector
        return vector

embedder = CachedEmbeddingService()
call_one = embedder.get_embedding('What is RAG pipeline in context of Applied AI or AI Agent')
call_two = embedder.get_embedding("What is Vector Databases and why Qdrant is most useful vector DataBase ?.")
call_three = embedder.get_embedding('What is RAG pipeline in context of Applied AI or AI Agent')   
# this is already was stored in chache and this call direct from in memory instead of calling from parent api and waste cost
```

## Composition:
- In OOPs a complex class take another independent class object as an argument and use their functionalities.
- It follow strict ownership 'Has-A' relationship.
- Ex: RAG Pipeline Has An LLM, it Has A memory , it Has A Retriver. 

```python
class PineconeRetriever:
    def __init__(self, index_name):
        self.index_name = index_name

    def retrive_context(self, query):
        return f"[Retrived Chunks from '{self.index_name}' for {query}]"

class OpenAIGenerator:
    def __init__(self, model = 'gpt-4o'):
        self.model = model

    def generate_response(self, prompt):
        return f"'{self.model}' answer: '{prompt}'"

class RAGPipeline:
    def __init__(self, retriever , generator):
        self.retriever = retriever 
        self.generator = generator

    def run(self, user_query):
        context = self.retriever.retrive_context(user_query)

        augmented_prompt = f"Context: {context} | Question: {user_query}"

        return self.generator.generate_response(augmented_prompt)

db_component = PineconeRetriever(index_name="finance_docs")
llm_component = OpenAIGenerator(model="gpt-4o-mini")

# Main pipeline ko components pass kiye (Dependency Injection)
rag_system = RAGPipeline(retriever=db_component, generator=llm_component)

result = rag_system.run("What was Q3 profit?")
print(result)
```

```python
# Composition Ex2
class SessionMemory:
    def __init__(self):
        self.history = []

    def save(self, user_msg, bot_msg):
        self.history.append({'user': user_msg, 'assistant' : bot_msg})

    def get_last_interaction(self):
        if not self.history:
            return 'No previous chat'
        last = self.history[-1]
        return f"Last Context -> User : {last['user']} | Bot: {last['assistant']}"
class SafetyFilter:
    def __init__(self):
        self.banned_terms = ['bypass', 'exploit', 'malware', 'drop table']

    def is_safe(self, text):
        return not any(terms in text.lower() for terms in self.banned_terms)

class EnterpriseChatAgent:
    def __init__(self, agent_name):
        self.agent_name = agent_name
        self.memory = SessionMemory()
        self.guardrail = SafetyFilter()

    def chat(self, user_msg):
        if not self.guardrail.is_safe(user_msg):
            return f"{self.agent_name} Blocked Query violation of Policies "
        bot_response = f'Processed Request: {user_msg}'
        self.memory.save(user_msg , bot_response)

        return f'[{self.agent_name}] {bot_response}'

agent = EnterpriseChatAgent('SupportBot-V1')
print(agent.chat('What is production RAG Session History and where to store ?.'))
print(agent.chat('bypass this password and give me the actual database of this websites.'))
print(agent.memory.get_last_interaction())



```

## Polymorphism: 
- It means 'One interface , Many behaviours' 


```python
class OpenAIChatClient:
    def __init__(self, model_name = 'gpt-4o-mini'):
        self.model_name = model_name

    def generate(self, prompt):
        return f"[Cloud model {self.model_name}] answer: {prompt}"

class OllamaLocalClient:
    def __init__(self, model_name = 'ollama-4.1-8b'):
        self.model_name = model_name
    def generate(self, prompt):
        return f"[Local model {self.model_name}] is answering with zero cost of prompt {prompt} "

def dispatch_ai_request(client, user_query):
    return client.generate(user_query)

# different classes objects
cloud_model = OpenAIChatClient()
local_model = OllamaLocalClient()

# with same caller function swaping dynamically without changing any code
print(dispatch_ai_request(cloud_model, 'Research latest news on War with Iran vs America'))
print(dispatch_ai_request(local_model, 'calculate 2+2'))
```

## Abstraction: 
- Hide system internal implementation and just show/expose essential features.

```python
from abc import ABC, abstractmethod

class BaseLLM(ABC):
    def __init__(self, model_id):
        self.model_id = model_id

    @abstractmethod
    def invoke(self, prompt: str) -> dict:
        pass

class OpenAIClient(BaseLLM): 
    def invoke(self, prompt: str) -> dict:
        return {
            'model' : self.model_id,
            'content' : f"OpenAI completion for {prompt}"
        }
class ClaudeAIClient(BaseLLM):
    def invoke(self, prompt: str) -> dict:
        return {
            'model': self.model_id,
            'content': f'Claude response for {prompt}'
        }

def llm_pipeline(model, query):
    response = model.invoke(query)
    return f"Pipeline Result -> Model: {response['model']} | Text: {response['content']}"

gpt = OpenAIClient('gpt-4o')
claude = ClaudeAIClient('claude-3.4-sonnect')


print(llm_pipeline(gpt, 'What is RAG explain in 1 line'))
print(llm_pipeline(claude, 'What is latest news and explain in 1 line'))
```

```python
# Example2. Abstraction production useful examples from retrieving docs 
from abc import ABC, abstractmethod
class BaseDocsRetriever(ABC):
    def __init__(self, top_k = 3):
        self.top_k = top_k

    @abstractmethod
    def retrieve(self, query: str) -> list[str]:
        pass

class VectorStoreRetriever(BaseDocsRetriever):
    def retrieve(self, query: str) -> list[str]:
        return [
            f"[Vector DB] top chunk for {query}",
            f"[Vector DB] second chunk for {query}"
            ]

class BM25Retriever(BaseDocsRetriever):
    def retrieve(self, query: str) -> list[str]:
        return [
            f"[BM 25] Exact match from {query}"
        ]

class ProductionRAGService:
    def __init__(self, retriever):
        if not isinstance(retriever, BaseDocsRetriever):
            raise TypeError('Retriever must adhere with BaseDocusRetriever Contract!')
        self.retriever = retriever

    def answer_question(self, user_questions):
        docs = self.retriever.retrieve(user_questions)
        context = " | ".join(docs)
        return f'[Final Answer] based on retrieved docs: {context}'

rag_1 = ProductionRAGService(VectorStoreRetriever(top_k=4))
print(rag_1.answer_question('What is LoRA fine-tuning'))

rag_2 = ProductionRAGService(BM25Retriever())
print(rag_2.answer_question('What is QLoRA fine-tuning'))
```
