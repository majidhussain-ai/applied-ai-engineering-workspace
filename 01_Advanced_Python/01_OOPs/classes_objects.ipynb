## Classes and Objects:
* Class  : Class is user-defined prototype or blueprint that define the attributes and methods of class.
* Objects  : It is physical, real world entity or instance of any class that occupied memory when it created.

```python
# Real world useful example of classes and objects.
class LLMConnector:
    def __init__(self, model_name, api_key, temperature = 0.7):
        self.model_name = model_name
        self.api_key = api_key
        self.temperature = temperature
        self.is_connected = False

    def connect(self):
        self.is_connected = True 
        return f'Connection established with {self.model_name}'

    def generate_text(self, prompt):
        if not self.is_connected:
            raise Exception('Model is not connected, first call connect()')

        return f"{self.model_name} at temp {self.temperature} is answering '{prompt}'"

open_ai_model = LLMConnector(model_name = 'GPT-4o', api_key = 'sk-123', temperature = 0.2)

open_ai_model.connect()

result = open_ai_model.generate_text('Explain RAG in one line concise and useful answer')
print(result)
```

```python
class OpenAIChat:
    total_instances = 0  # Class variable (Shared by all objects)

    def __init__(self, model_name: str):
        self.model_name = model_name
        OpenAIChat.total_instances += 1

    # @classmethod
    def get_total_instances(self):
        # cls.total_instances poori class ka shared data read karega
        return f"Total active models: {self.total_instances}"

# Object banaye
m1 = OpenAIChat("gpt-4o")
m2 = OpenAIChat("gpt-4o-mini")

# Direct Class ke naam se call karein
print(m2.get_total_instances())  # Output: Total active models: 2
```

```python
# Class method example
class LLMConfig:
    def __init__(self, name, provider, cost_per_token):
        self.name = name
        self.provider = provider
        self.cost_per_token = cost_per_token

    @classmethod
    def from_config(cls, raw_text):
        parts = raw_text.split('-')

        if len(parts) < 3:
            raise ValueError("Invalid format. Expected 'name-provider-cost_per_token'")

        # Example: "GPT-4o-OpenAI-0.002"
        # - last part = cost
        # - second last = provider
        # - everything before = model name
        cost_per_token = float(parts[-1])
        provider = parts[-2]
        name = '-'.join(parts[:-2])

        return cls(name, provider, cost_per_token)

# normal way
config1 = LLMConfig('claude-3', 'Anthropic', 0.015)

# Alternative constructor method
config2 = LLMConfig.from_config("GPT-4o-OpenAI-0.002")
print(config2.name, config2.provider, config2.cost_per_token)
```

```python
# Samve as above but with different string input
class LLMConnect:
    def __init__(self,name,provider ,cost_per_token):
        self.name = name
        self.provider = provider
        self.cost_per_token = cost_per_token


    @classmethod
    def from_string(cls, raw_string):
        name , provider, cost_per_token = raw_string.split(',')
        return cls(name, provider, cost_per_token)

config = LLMConnect.from_string('GPT-4o,OpenAI, 0.002')
print(config.name, config.provider, config.cost_per_token)
```

## Instance vs Class vs static methods with production use-cases:

```python
# Instance Method:
class AIChatBot:
    def __init__(self, user_name):
        self.user_name = user_name
        self.chat_history = []
    #   This is isntance method
    def add_message(self, role, text):
        log_entry = f'{role} : {text}'
        self.chat_history.append(log_entry)
        print(f"[History updated for {self.user_name}] : {log_entry}")

user_a_bot = AIChatBot('Majid')
user_b_bot = AIChatBot('Amjad')

user_a_bot.add_message('user', 'What is LLMs')
user_b_bot.add_message('assistance', 'LLMs are massive foundation AI Models that based on Transfor Architecture')

print(f"Majid blogs : {user_a_bot.chat_history}")
print(f"Amjad blogs : {user_b_bot.chat_history}")
```

```python
# instance method example3: token limiter counter agent
class TokenBucket:
    def __init__(self, max_tokens):
        self.max_tokens = max_tokens
        self.current_tokens = max_tokens

    def consume_tokens(self, count):
        if self.current_tokens >= count:
            self.current_tokens -= count
            print(f'You have consumed tokens {count} , Remaining tokens are : {self.current_tokens}')
            return True
        print('Error: Rate limit exceeded! ')
        return False

bucket = TokenBucket(100)
bucket.consume_tokens(30)
bucket.consume_tokens(80)
```

```python
# classmethod code example
class ModelConfig:
    def __init__(self, model_name, max_tokens , temperature):
        self.model_name = model_name
        self.max_tokens = max_tokens
        self.temperature = temperature

    @classmethod
    def from_dict(cls, raw_json):
        return cls(
            model_name=raw_json.get('model','gemini-3-pro'),
            max_tokens = raw_json.get('max_tokens', 1556),
            temperature = raw_json.get('temp', 0.7)
        )

raw_json_data = {'model_name':'GPT-40','max_tokens':6500, 'temp':0.3 }
config = ModelConfig.from_dict(raw_json_data)
print(config.model_name)
print(config.temperature)
print(config.max_tokens)


```

```python
class AIModel:
    def __init__(self, model_name , temperature):
        self.model_name = model_name
        self.temperature = temperature

    def generate_text(self, prompt):
        return (f"The model '{self.model_name}' is used with temp '{self.temperature}' to generate answer of '{prompt}'")

    @classmethod
    def default_model(cls):
        return cls(
            model_name = 'gpt-4o',
            temperature = 0.2
        )

    @staticmethod
    def validate_prompt(prompt):
        clean_prompt = prompt.strip()
        if not clean_prompt:
            raise ValueError("prompt can not be empty")
        return clean_prompt.lower()

model = AIModel('claude-3-sonnect', 0.3)
print(model.generate_text('What is Artificial Intelligence'))

default_model = AIModel.default_model()
print(default_model.model_name)

print(default_model.temperature)

valide_prompt = AIModel.validate_prompt('')
valide_prompt
```

## Dunder/Magic methods:
- __init__ — object initialization
- __repr__ — debugging, logging, observability
- __str__ — human-readable representation
- __eq__ — object equality/comparison
- __hash__ — hashing, sets, dictionary keys, caching
- __iter__ — custom iterables, streaming/data pipelines
- __getitem__ — list/dict-like custom objects
- __call__ — callable objects; very relevant to AI components
- __enter__ — context manager setup
- __exit__ — context manager cleanup

```python
class MyBot:
    def __init__(self, ai_answer, model):
        self.ai_answer = ai_answer
        self.model = model

    def __str__(self):
        return f"AI says : {self.ai_answer}"

    def __repr__(self):
        return f"MyBot(ai_answer = {self.ai_answer} , model_version = {self.model})"

debug_bot = MyBot('RAG uses external knowledge to give accurate answer to LLMs','gpt-4o-v2')

# normal print
print(debug_bot)

# developer display
print([debug_bot])
```

```python
# 4 dunder __eq__ with code example
# RAG chunks de-duplication checking

class VectorChunk:
    def __init__(self, chunk_id , content):
        self.chunk_id = chunk_id
        self.content = content

    def __eq__(self, other):
        if not isinstance(other, VectorChunk):
            return False
        return self.chunk_id == other.chunk_id and self.content == other.content

chunk_a = VectorChunk('id_01', 'Applied AI is growing fast in 2026')
chunk_b = VectorChunk('id_01', 'Applied AI is growing fast in 2026')

print(chunk_a == chunk_b)
```
