### Iterable: 
- The object that holds the data (like a list, dictionary, or string).
### Iterator:
- The hidden helper that Python calls by default when you use a for loop to grab one item at a time.

### Iteration:
- The whole process of working from start to end.

```python
# Custom range iterator
class StepCounter:
    def __init__(self, start, end , step = 2):
        self.current = start
        self.end = end
        self.step = step

    # always iter return itself 
    def __iter__(self):
        return self

    def __next__(self):
        if self.current >= self.end:
            raise StopIteration
        value_return = self.current
        self.current += self.step
        return value_return

counter = StepCounter(start=1, end = 20, step=5)
for count in counter:
    print(count)
print('End of counter function')
```

```python
# RAG chunker hunlder
class RAGChunkBatch:
    def __init__(self, documents, chunk_size = 2):
        self._documents = documents
        self._chunk_size = chunk_size
        self._index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self._index >= len(self._documents):
            raise StopIteration

        batch = self._documents[self._index: self._index + self._chunk_size]
        self._index += self._chunk_size
        return batch

simulated_rag_documents = [
    "Doc 1: LangGraph maintains stateful multi-agent graphs.",
    "Doc 2: Qdrant indexes dense vectors using HNSW graphs.",
    "Doc 3: Transformers rely on multi-head scaled dot-product attention.",
    "Doc 4: LoRA injects low-rank trainable decomposition matrices.",
    "Doc 5: Pydantic enforces data contracts on structured LLM outputs."
]

batch_iterator = RAGChunkBatch(simulated_rag_documents, chunk_size=2)

batch_number = 1
for chunk_batch in batch_iterator:
    print(f"---> Processing batch #{batch_number} (size = {len(chunk_batch)})")
    for doc in chunk_batch:
        print(f"---Embedding: {doc[:40]}.....")
        batch_number += 1
```

```python
class RAGChunkBatch:
    def __init__(self, documents, chunk_size = 2):
        self._document = documents
        self._chunk_size = chunk_size
        self._index = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self._index >= 0:
            raise StopIteration
        # slicing current batch
        batch  = self._document[self._index : self._index + self._chunk_size]
        self._index += self._chunk_size
        return batch

simulated_rag_documents = [
    "Doc 1: LangGraph maintains stateful multi-agent graphs.",
    "Doc 2: Qdrant indexes dense vectors using HNSW graphs.",
    "Doc 3: Transformers rely on multi-head scaled dot-product attention.",
    "Doc 4: LoRA injects low-rank trainable decomposition matrices.",
    "Doc 5: Pydantic enforces data contracts on structured LLM outputs."
]

batch_iterator = RAGChunkBatch(simulated_rag_documents, chunk_size=2)

batch_number = 1
for chunk_batch in batch_iterator:
    print(f"----> Processing batch #{batch_number} (size: {len(chunk_batch)})")
    for docs in chunk_batch:
        print(f"---Embeddiing")
```

## Generator:
- Generator is a special function in python that automatically implements iterator protocols without making any class poilerpoint.
- When any function returning YIELD keword that is not normal function , then it is generator function

```python
# simple example with manually understanding yield keyword

def token_stepper():
    print('Stage 1: Starting Function......')
    yield 'Return Alpha'

    print("Stage 2: Resumed after first yield....")
    yield 'Retun Beta'

    print("Stage 3: Resumed After second yield....")
    yield 'Return Gamma'
    print("Stage 4: Reached end of function.")

gen = token_stepper()

print("----First Call------")
print("Received:" , next(gen))

print("\n----Second Call------")
print("Received:" , next(gen))

print("\n----Third Call------")
print("Received:" , next(gen))
try: 
    print("\n----Fourth Call------")
    print("Received:" , next(gen))
except StopIteration:
    print("\n[Exhausted] Generator successfully completed StopIteration")
```

```python
# Memory comparison list vs generator
import sys
import time
def vector_chunks_list(n):
    start = time.perf_counter()
    data = [
        f"Vector chunks index #{i} with semantic text embedding"
        for i in range(n)
    ]
    end = time.perf_counter()
    execution_time = end - start
    return data, execution_time

def vector_chunks_gen(n):
    for i in range(n):
        yield f"Vector chunks index #{i} with semantic text embedding"

chunks_items = 5000000

list_data, list_time = vector_chunks_list(chunks_items)
gen_data = vector_chunks_gen(chunks_items)

list_in_bytes = sys.getsizeof(list_data)
gen_in_bytes = sys.getsizeof(gen_data)

print(f"Time taken by list comprehension: {list_time:.3f} seconds")
print(f"\nMemory used by list:  {list_in_bytes / (1024 *1024) :.2f} MB")
print(f"Memory used by gen:  {gen_in_bytes} Bytes (~{gen_in_bytes/1024: .2f}) KB")
```

```python
import time
from typing import Generator

def stream_llm_response(prompt: str) -> Generator[str, None, None]:
    """
    Simulates token-by-token streaming from an LLM API 
    (equivalent to OpenAI client.chat.completions.create(stream=True)).
    """
    simulated_response = (
        "Retrieval-Augmented Generation (RAG) reduces hallucinations "
        "by grounding large language models on external enterprise facts."
    )
    tokens = simulated_response.split(" ")

    for token in tokens:
        time.sleep(0.08)  
        yield token + " "  # Realtime yield to the client

print(f"User Prompt: 'Explain RAG benefits'\n")
print("Streaming Assistant Response: ", end="", flush=True)

# Consumer loop: Har yielded token aate hi print ho raha hai
start_stream = time.perf_counter()
total_tokens = 0

for token_chunk in stream_llm_response("Explain RAG benefits"):
    print(token_chunk, end="", flush=True)
    total_tokens += 1

duration = time.perf_counter() - start_stream
print(f"\n\n[Telemetry] Streamed {total_tokens} tokens in {duration:.2f}s (~{total_tokens/duration:.1f} tokens/sec)")
```
