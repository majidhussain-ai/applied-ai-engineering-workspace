## Exception Handling:
* Exception handling is a mechanism in Python used to handle runtime errors gracefully so the application does not crash. We use a try-except block to catch unexpected errors (like missing data, database downtime, or network failures). In production AI, this guarantees fault tolerance, allowing the system to log the error, alert developers, or fallback to a safe option without shutting down the entire server.

### Common Built-in Errors in Python:
- `SyntaxError` – Invalid Python syntax.
- `NameError` – Using a variable that is not defined.
- `TypeError` – An operation uses an inappropriate data type.
- `ValueError` – A function receives an invalid value.
- `IndexError` – A sequence index is out of range.
- `KeyError` – A dictionary key does not exist.
- `AttributeError` – An object does not have the requested attribute.
- `ZeroDivisionError` – Division or modulo by zero.
- `FileNotFoundError` – A requested file does not exist.
- `ImportError` – A module or name cannot be imported.
- `ModuleNotFoundError` – The specified module cannot be found.
- `PermissionError` – Insufficient permission for an operation.
- `OverflowError` – A numerical result is too large.
- `RecursionError` – Maximum recursion depth is exceeded.
- `IndentationError` – Incorrect indentation.

```python
# simple example with except block: checking eligibilty erea of any adult who eligible of card or license
# we just use try except block where we perform or calling to action of program not where we're building code from sratch

def eligible_critera(age):
    if age < 18:
        raise ValueError("You are not eligible for license")
    print("Congratulations!, You're eligible for license")
try: 
    check_critera = eligible_critera(14)
    print(check_critera)
except ValueError as e:
    print(e)
```

```python
def process_user_prompt(prompt: str, max_words: int = 100):
    if not isinstance(prompt , str):
        raise TypeError(f'Expected string prompt, got {type(prompt).__name__}')
    
    words = prompt.split()
    if len(words) > max_words:
        raise ValueError(f"Prompt length ({len(words)} words) exceeds max words ({max_words} words)")

    return f"Processing prompt: {prompt}"

try:
    result = process_user_prompt(['majd', 'hussain'], max_words=50)

except (TypeError, ValueError) as e:
    print(f"Validation guardrail trigger: {e}")
```

```python
# calling 2nd time with exceeding prompt length
try:
    result = process_user_prompt('What is RAG in Agentic pipeline ?.', max_words=5)
    print(result)

except (TypeError, ValueError) as e:
    print(f"Processing guardrail triggered: {e}")
```

```python
import os
from pathlib import Path
class ModelConfigError(Exception):
    pass 

def load_llm_config(file_path):
    path = Path(file_path)

    try:
        with open(path, 'r') as file:
            return file.read()
    except FileNotFoundError as err:
        # this below line is concept chaining by using (from err) it mean from low-level error to join domain 
        raise ModelConfigError(f'Critical configuration missing at {path.name}') from err

    except PermissionError as err:
        raise ModelConfigError(f"Access denied to system file {path.name} ")

file_path = os.path.join(os.getcwd(), 'config_model.txt')  # Simulate making a path
try:
    load_model = load_llm_config(file_path)
    print(load_model)
except ModelConfigError as err:
    print(f'Handled error: {err}')
    print(f"Original error that occured: {repr(err.__cause__)}")
```
