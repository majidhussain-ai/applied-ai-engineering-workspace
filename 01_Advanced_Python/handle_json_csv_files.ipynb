```python
file_path = "/home/majid-hussain-ai/Desktop/applied-ai-engineering-workspace/01_Advanced_Python/rag_documents/text/test.csv"
logger = logging.getLogger('CSVReadFile')

def stream_csv_dataset(file_path):
    if not file_path.is_file():
        raise FileNotFoundError(f"CSV DataSet is missing: {file_path}")
    with open(file_path, mode='r', encoding='utf-8', newline='') as file:
        reader = csv.DictReader(file)
        if not reader.fieldnames:
            logger.warning(f"CSV File {file_path[-1]} is empty or has no header")
            return


        for row_idx, row in enumerate(reader, start=1):
            yield row
        
result = stream_csv_dataset(file_path)
print(result)
```

```python
base_dir = Path('ai_pipeline_run')
config_file_path = base_dir/'configs'/'model_configs.json'
log_file_path = base_dir/'logs'/'predictions.csv'

config_file_path.parent.mkdir(parents=True, exist_ok=True)
log_file_path.parent.mkdir(parents=True, exist_ok=True)

model_config = {
    'model_name' : 'ollama-3-8b',
    'temperature' : 0.7,
    'max_tokens' : 512,
    'version' : '1.0.0'
}

json_text = json.dumps(model_config)
config_file_path.write_text(json_text, encoding='utf-8')

print(f"Json config written to: {config_file_path}")
```
